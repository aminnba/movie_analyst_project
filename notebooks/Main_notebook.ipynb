```python
# Data structuring:
# 1) Remove old parsed columns if they exist
# 2) Parse genres into list + extract main genre
# 3) Parse cast into list of names + decode unicode + extract top 3 actors
# 4) Parse production countries into clean list of country names

import ast

# 1) Remove old parsed columns if they exist
df1 = df1.drop(columns=['genres_list', 'main_genre', 'cast_list', 'top_3_cast', 'countries_list'], errors='ignore')


# 2) Genres → list + main genre
def parse_genres(g):
    if pd.isna(g):
        return None
    
    genres = g.split()
    cleaned = []
    i = 0

    while i < len(genres):
        # Merge "Science Fiction" into one token
        if i < len(genres)-1 and genres[i] == "Science" and genres[i+1] == "Fiction":
            cleaned.append("Science Fiction")
            i += 2
        else:
            cleaned.append(genres[i])
            i += 1
            
    return cleaned

df1['genres_list'] = df1['genres'].apply(parse_genres)
df1['main_genre'] = df1['genres_list'].apply(lambda x: x[0] if isinstance(x, list) and len(x) > 0 else None)


# 3) Cast → list + unicode decode + top 3 actors
def parse_cast_names(cast_str):
    if pd.isna(cast_str):
        return None
    
    parts = cast_str.split()
    names = []
    i = 0
    suffixes = {"Jr.", "Sr.", "II", "III", "IV"}

    while i < len(parts) - 1:
        name = parts[i] + " " + parts[i+1]
        i += 2

        if i < len(parts) and parts[i] in suffixes:
            name += " " + parts[i]
            i += 1

        names.append(name)

    return names

df1['cast_list'] = df1['cast'].apply(parse_cast_names)

def decode_unicode_list(names):
    if not isinstance(names, list):
        return names
    decoded = []
    for n in names:
        try:
            decoded.append(bytes(n, 'utf-8').decode('unicode_escape'))
        except:
            decoded.append(n)
    return decoded

df1['cast_list'] = df1['cast_list'].apply(decode_unicode_list)
df1['top_3_cast'] = df1['cast_list'].apply(lambda x: x[:3] if isinstance(x, list) else None)


# 4) Production countries → list of names only
def parse_countries(country_str):
    if pd.isna(country_str):
        return None
    try:
        country_data = ast.literal_eval(country_str)
        return [c['name'] for c in country_data if 'name' in c]
    except:
        return None

df1['countries_list'] = df1['production_countries'].apply(parse_countries)


# Preview
df1[['genres_list', 'main_genre', 'cast_list', 'top_3_cast', 'countries_list']].head(10)


```

```python
# Data quality flags (idempotent):
# 1) Remove existing flags if re-running
# 2) Flag movies missing critical financial fields (budget or revenue)
# 3) Flag suspicious runtimes (<45 or >300 minutes)
# 4) Flag financial outliers based on revenue IQR
# 5) Flag invalid budgets (<$10,000)
# 6) Summary with counts and percentages

flag_cols = ['missing_critical', 'runtime_suspicious', 'financial_outlier', 'invalid_budget']
df1.drop(columns=[c for c in flag_cols if c in df1.columns], inplace=True, errors='ignore')

# 2) Missing critical fields
df1['missing_critical'] = df1[['budget', 'revenue']].isna().any(axis=1)

# 3) Suspicious runtimes
df1['runtime_suspicious'] = df1['runtime'].apply(
    lambda x: True if pd.notna(x) and (x < 45 or x > 300) else False
)

# 4) Financial outliers (revenue IQR)
Q1 = df1['revenue'].quantile(0.25)
Q3 = df1['revenue'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df1['financial_outlier'] = df1['revenue'].apply(
    lambda x: True if pd.notna(x) and (x < lower_bound or x > upper_bound) else False
)

# 5) Invalid budgets
df1['invalid_budget'] = df1['budget'].apply(
    lambda x: True if pd.notna(x) and x < 10000 else False
)

# 6) Summary
data_quality_summary = {
    'missing_critical (%)': round(df1['missing_critical'].mean() * 100, 2),
    'runtime_suspicious (%)': round(df1['runtime_suspicious'].mean() * 100, 2),
    'financial_outlier (%)': round(df1['financial_outlier'].mean() * 100, 2),
    'invalid_budget (%)': round(df1['invalid_budget'].mean() * 100, 2)
}
data_quality_summary
```

```python
# ============================
# KPI Engineering (Idempotent)
# ============================

# 1) Remove existing KPI columns if re-running
kpi_cols = ['profit', 'roi', 'profit_margin', 'audience_engagement', 'success_flag']
df1.drop(columns=[c for c in kpi_cols if c in df1.columns], inplace=True, errors='ignore')

# 2) Core KPI definitions:
# profit            = revenue - budget
# roi               = (revenue - budget) / budget
# profit_margin     = profit / revenue
# audience_engagement = vote_count * vote_average (proxy for cultural engagement)
# success_flag      = 1 if movie made profit, else 0

# 3) Compute KPIs
df1['profit'] = df1['revenue'] - df1['budget']

df1['roi'] = df1.apply(
    lambda x: (x['profit'] / x['budget']) if pd.notna(x['budget']) and x['budget'] > 0 else None, axis=1
)

df1['profit_margin'] = df1.apply(
    lambda x: (x['profit'] / x['revenue']) if pd.notna(x['revenue']) and x['revenue'] > 0 else None, axis=1
)

df1['audience_engagement'] = df1.apply(
    lambda x: x['vote_count'] * x['vote_average'] if pd.notna(x['vote_count']) and pd.notna(x['vote_average']) else None,
    axis=1
)

df1['success_flag'] = df1.apply(
    lambda x: 1 if pd.notna(x['profit']) and x['profit'] > 0 else 0,
    axis=1
)

# 4) KPI Summary Preview
kpi_summary = {
    'Total Movies': len(df1),
    'Profitable Movies (%)': round(df1['success_flag'].mean() * 100, 2),
    'Avg ROI': round(df1['roi'].mean(), 2),
    'Avg Profit ($)': round(df1['profit'].mean(), 2),
}
kpi_summary

# Review KPI summary:
# the average return is +2,532× the budget.
#This tells us outliers are hugely distorting the averages.

```

```python
# KPI summary (cleaned dataset):
# 1) Exclude financial outliers and invalid budgets
# 2) Compute KPIs on the cleaned dataset

df_clean = df1[
    (df1['financial_outlier'] == False) &
    (df1['invalid_budget'] == False)
].copy()

kpi_summary_clean = {
    'Total Movies': len(df_clean),
    'Profitable Movies (%)': round(df_clean['success_flag'].mean() * 100, 2),
    'Avg ROI': round(df_clean['roi'].mean(), 2),
    'Median ROI': round(df_clean['roi'].median(), 2),
    'Avg Profit ($)': round(df_clean['profit'].mean(), 2),
}
kpi_summary_clean
```

```python
# Light cleaning for IMDb dataset (df2):
#Note:
# The IMDb dataset (df2) is cleaned only at this stage.
# As a secondary source, only relevant columns and merge keys are standardized
# to ensure efficiency and avoid unnecessary preprocessing of unused data.

# 1) Keep only relevant columns for merge and analysis
# 2) Standardize text and numeric fields
# 3) Handle missing or invalid rating values
# 4) Quick summary check

# =============================
# IMDb Data Reset + Alignment (final, consistent)
# =============================

# 1) Reload the IMDb dataset fresh
df2 = pd.read_csv(path2)

# 2) Keep only relevant columns
imdb_cols = ['title', 'year', 'imdbRating', 'imdbVotes', 'metacritic', 'writer', 'country']
df2 = df2[imdb_cols].copy()

# 3) Standardize text fields
df2['title'] = df2['title'].astype(str).str.strip().str.lower()
df2['writer'] = df2['writer'].astype(str).str.strip()
df2['country'] = df2['country'].astype(str).str.strip()

# 4) Convert numeric fields
df2['imdbRating'] = pd.to_numeric(df2['imdbRating'], errors='coerce')
df2['imdbVotes']  = pd.to_numeric(df2['imdbVotes'], errors='coerce')
df2['metacritic'] = pd.to_numeric(df2['metacritic'], errors='coerce')

# 5) Extract year for merge and cast to Int64 (nullable int)
df2['release_year'] = pd.to_numeric(df2['year'], errors='coerce').astype('Int64')

# =============================
# Main Dataset Alignment
# =============================

# Ensure release_year exists in df_clean and is also Int64
df_clean['release_year'] = pd.to_datetime(df_clean['release_date'], errors='coerce').dt.year.astype('Int64')

# Standardize title
df_clean['title'] = df_clean['title'].astype(str).str.strip().str.lower()

# =============================
# Merge on title + release_year
# =============================

df_merged = pd.merge(
    df_clean,
    df2[['title', 'release_year', 'imdbRating', 'imdbVotes', 'metacritic', 'writer', 'country']],
    on=['title', 'release_year'],
    how='left'
)

# =============================
# Merge Coverage Summary
# =============================

coverage_summary = {
    'Total Movies': len(df_merged),
    'Matched with IMDb (%)': round(df_merged['imdbRating'].notna().mean() * 100, 2),
    'Metacritic Coverage after Merge (%)': round(df_merged['metacritic'].notna().mean() * 100, 2)
}
coverage_summary
```

```python
# ===========================================
# Critics vs Financial Performance (Phase 2)
# ===========================================

# 1) Prepare working subset
#    Keep only movies that have both financial & Metacritic data
critics_fin = df_merged[
    df_merged['metacritic'].notna() & df_merged['roi'].notna()
].copy()

# 2) Basic sanity check
print("Movies included in analysis:", len(critics_fin))

# 3) Correlation between Metacritic and ROI
# Winsorize ROI at 99th percentile
roi_cap = critics_fin['roi'].quantile(0.99)
critics_fin['roi_winsor'] = np.where(critics_fin['roi'] > roi_cap, roi_cap, critics_fin['roi'])

corr_value = critics_fin['metacritic'].corr(critics_fin['roi_winsor'])
print("Correlation (Metacritic vs ROI):", round(corr_value, 3))

# 4) Scatter plot
plt.figure(figsize=(8,6))
plt.hexbin(critics_fin['metacritic'], critics_fin['roi_winsor'], 
           gridsize=30, cmap='viridis', mincnt=1)
plt.colorbar(label='Movie Count')
plt.title("Critic Score vs ROI (Density View)")
plt.xlabel("Metacritic Score")
plt.ylabel("ROI (trimmed at 99th percentile)")
plt.ylim(0, 20)
plt.show()

# 5) Quick aggregation by critic score bucket
critics_fin['metacritic_bin'] = pd.cut(
    critics_fin['metacritic'],
    bins=[0, 20, 40, 60, 80, 100],
    labels=['0–20', '21–40', '41–60', '61–80', '81–100']
)
critics_summary = critics_fin.groupby('metacritic_bin').agg(
    avg_roi_raw=('roi','mean'),
    avg_roi_winsor=('roi_winsor','mean'),
    median_roi=('roi','median'),
    movie_count=('title','count')
).reset_index()

print(critics_summary)

# Review observations:
# - Winsorizing capped extreme ROI values to stabilize averages.
# - Correlation quantifies overall critics vs profit alignment.
# - Hexbin plot visualizes density rather than individual overlap.
# - Grouped summary shows profitability trends across critic tiers.


```

```python
corr_raw = critics_fin['metacritic'].corr(critics_fin['roi'])
corr_winsor = critics_fin['metacritic'].corr(critics_fin['roi_winsor'])

print(f"Raw correlation (Metacritic vs ROI): {round(corr_raw, 3)}")
print(f"Winsorized correlation (Metacritic vs ROI): {round(corr_winsor, 3)}")

#The raw correlation between critic scores and ROI was close to zero (0.02), 
# indicating no clear link when outliers are included.
# After capping extreme ROI values (winsorization), the correlation rose to 0.19, 
# Modest but consistent positive relationship, suggesting that well-reviewed films are more likely to achieve healthy returns, though commercial success remains highly variable.
```
